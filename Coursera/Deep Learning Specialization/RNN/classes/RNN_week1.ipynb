## 📚 Backpropagation Through Time (BPTT)

O **Backpropagation Through Time** é uma extensão do algoritmo de backpropagation para redes recorrentes (RNNs). Ele calcula gradientes considerando dependências temporais, "desdobrando" a rede ao longo do tempo.

- Cada passo temporal $ t $ calcula:
  - Estado oculto: $ a^{\langle t \rangle} $
  - Saída: $ \hat{y}^{\langle t \rangle} $
  - Perda: $ \mathcal{L}^{\langle t \rangle} $

- Pesos envolvidos:
  - $( W_{aa} $): recorrente
  - $( W_{ax} $): entrada
  - $( W_{ya} $): saída

### Perda total:
$$
\mathcal{L}(\hat{y}, y) = \sum_{t=1}^{T_{op}} \mathcal{L}^{\langle t \rangle}(\hat{y}^{\langle t \rangle}, y^{\langle t \rangle})
$$

### Exemplo de perda binária:
$$
\mathcal{L}^{\langle t \rangle} = - y^{\langle t \rangle} \log \hat{y}^{\langle t \rangle} - (1 - y^{\langle t \rangle}) \log (1 - \hat{y}^{\langle t \rangle})
$$
