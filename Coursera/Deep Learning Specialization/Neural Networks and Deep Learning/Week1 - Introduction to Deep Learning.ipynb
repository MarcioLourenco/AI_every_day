## 🧠 Conceito Básico de Rede Neural

- Deep Learning treina **redes neurais**, que mapeiam entradas \(x\) (como características de uma casa) para saídas \(y\) (como o preço).  
- Cada **neurônio** aplica uma função linear e uma ativação **ReLU**, por exemplo, que zera valores negativos.  
- Ao empilhar vários neurônios em **camadas**, a rede aprende automaticamente representações internas e faz previsões precisas.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Exemplo simples: prever preço com base no tamanho da casa

def relu(x):
    return np.maximum(0, x)
```

```python
# Define o domínio
x = np.linspace(-10, 10, 200)

y = relu(x)

# Plot
plt.plot(x, y)
plt.title("Função ReLU (Rectified Linear Unit)")
plt.xlabel("x")
plt.ylabel("ReLU(x)")
plt.grid(True)
plt.show()
```

```python
# pesos e bias (treinados manualmente aqui só para ilustrar)
w = 0.3   # peso
b = -50   # bias

# tamanho da casa (em m2)
x = np.array([100, 150, 200, 250])

# saída da rede (predição do preço)
y_pred = relu(w * x + b)

print("Preços previstos:", y_pred)
```

## Exemplo - Predição dos preços das casas - Redes Neurais

```python

# Posições dos nós
input_nodes = ['size', 'bedrooms', 'zip code', 'wealth']
hidden_nodes = ['h1', 'h2', 'h3']
output_nodes = ['price']

# Coordenadas (x, y)
positions = {
    'size': (1, 3),
    'bedrooms': (1, 2),
    'zip code': (1, 1),
    'wealth': (1, 0),
    'h1': (2, 2.5),
    'h2': (2, 1.5),
    'h3': (2, 0.5),
    'price': (3, 1.5)
}

# Cria figura
fig, ax = plt.subplots(figsize=(6, 4))

# Ligações: totalmente conectadas (input -> hidden)
for i in input_nodes:
    for h in hidden_nodes:
        ax.plot(
            [positions[i][0], positions[h][0]],
            [positions[i][1], positions[h][1]],
            'k-', alpha=0.5
        )

# Ligações: hidden -> output
for h in hidden_nodes:
    ax.plot(
        [positions[h][0], positions['price'][0]],
        [positions[h][1], positions['price'][1]],
        'k-', alpha=0.5
    )

# Desenha nós
for node, (x, y) in positions.items():
    ax.scatter(x, y, s=1000, color='white', edgecolors='black', zorder=3)
    ax.text(x, y, node, ha='center', va='center', fontsize=10)

# Ajustes visuais
ax.axis('off')
plt.title("Housing Price Prediction Neural Network", fontsize=12)
plt.show()
```

### 🤖 Supervised Learning e Aplicações de Redes Neurais

- A maior parte do valor econômico das redes neurais vem do **aprendizado supervisionado**, onde o modelo aprende uma função que mapeia **entradas (x)** para **saídas (y)**.  
- Exemplos:  
  - Predição de **preço de casas**  
  - **Propaganda online** (probabilidade de clique)  
  - **Visão computacional** (identificação de imagens)  
  - **Reconhecimento de fala** (áudio → texto)  
  - **Tradução automática** (inglês → chinês)  
  - **Direção autônoma** (imagem + radar → posição de carros)  

- Tipos de redes:  
  - **Rede neural padrão (DNN)** → dados estruturados (ex: imóveis, anúncios)  
  - **CNN** → imagens  
  - **RNN** → sequências (áudio, texto)  
  - **Híbridas** → tarefas complexas (ex: direção autônoma)  

- **Dados estruturados:** tabelas com colunas bem definidas (ex: tamanho, idade, preço).  
- **Dados não estruturados:** imagens, texto, áudio.  

💡 *Deep Learning tornou possível para computadores entenderem dados não estruturados, criando valor tanto em aplicações “visuais” quanto em sistemas baseados em dados estruturados corporativos.*

# 🚀 Por que o Deep Learning decolou

## 1. Dados
- Mais dados digitais (apps, sensores, IoT).  
- Modelos clássicos saturam com muitos dados.  
- Redes neurais continuam melhorando.

## 2. Computação
- GPUs/TPUs → redes maiores e mais rápidas.  
- Possibilitou treinar modelos enormes.  

## 3. Algoritmos
- Troca da **Sigmoid → ReLU** acelerou o aprendizado.  
- ReLU evita gradiente próximo de zero.  

## 4. Escala
- Escala = dados + modelo + hardware.  
- “Scale drives progress.”  

## 5. Ciclo de experimentação


```python
import numpy as np
import matplotlib.pyplot as plt

# Eixo X: quantidade de dados
x = np.linspace(0, 10, 200)

# Curvas representando diferentes algoritmos
svm = 0.6 * (1 - np.exp(-0.8 * x))
small_nn = 0.8 * (1 - np.exp(-0.4 * x))
large_nn = 1.2 * (1 - np.exp(-0.15 * x))

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, svm, label="Modelos Tradicionais (SVM, Regressão)", linestyle='--')
plt.plot(x, small_nn, label="Rede Neural Pequena", linestyle='-.')
plt.plot(x, large_nn, label="Rede Neural Grande", linewidth=2)

plt.title("Desempenho vs Quantidade de Dados", fontsize=14)
plt.xlabel("Quantidade de Dados (m)")
plt.ylabel("Desempenho do Modelo")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
